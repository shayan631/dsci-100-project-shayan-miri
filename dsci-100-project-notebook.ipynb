#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.

*You may add more code cells if needed.*

```r
library(rsample)
library(tidymodels)
# setting seed for consistent results
set.seed(67575209)

# making initial split for training and testing data sets
basketball_split <- initial_split(basketball_condensed_finish, prop = 0.75, strata = finish_condensed)
basketball_training <- training(basketball_split)
basketball_testing <- testing(basketball_split)

# v fold for cross validation
basketball_fold <- vfold_cv(data = basketball_training, v = 5, strata = finish_condensed)

# Making recipe and preprocessing steps
basketball_recipe <- recipe(finish_condensed ~ reg_wins_pct + conf_wins_pct, data = basketball_training) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors())

# values of k used for finding the best one
k_grid <- tibble(neighbors = seq(from = 1, to = 100, by = 5))

knn_tune <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |> set_engine("kknn") |> set_mode("classification")

knn_results <- workflow() |> add_recipe(basketball_recipe) |>
    add_model(knn_tune) |>
    tune_grid(resamples = basketball_fold, grid = k_grid) |>
    collect_metrics()

# the k with the highest accuracy
knn_best <-  knn_results |> filter(.metric == "accuracy") |> arrange(desc(mean)) |> slice(1) |> select(neighbors) |> pull()

# training model on the training data
knn_spec_best <- nearest_neighbor(weight_func = "rectangular", neighbors = knn_best) |>
    set_engine("kknn") |>
    set_mode("classification")
basketball_fit <- workflow() |> add_recipe(basketball_recipe) |> add_model(knn_spec_best) |> fit(basketball_training)

# testing model on the testing data
basketball_predictions <- predict(basketball_fit, basketball_testing) |>
    bind_cols(basketball_testing)

# metrics
basketball_predictions |>
    metrics(truth = finish_condensed, estimate = .pred_class) |>
    filter(.metric == "accuracy")

# confusion matrix
basketball_predictions |>
    conf_mat(truth = finish_condensed, estimate = .pred_class)
```
